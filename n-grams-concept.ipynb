```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({'text' : ['people watch campusx', 'campusx watch campusx', 'people write comment', 'campusx write comment'], 'output' : [1,1,0,0]})
```

```python
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(ngram_range=(1,3))
```

```python
bow = cv.fit_transform(df['text'])
```

```python
# printing the vocabulary (unique words present in the corpus)

print(cv.vocabulary_)
```

```python
len(cv.vocabulary_)
```

```python
print(bow[0].toarray())    # 1st sentence
print(bow[1].toarray())    # 2nd sentence

# so we can see our output as a vector below
```

```python
# take a new sentence and predict for it :-

cv.transform(["campusx watch and write comment of campusx"]).toarray()
```

<h3> By changing one of the parameters in the sklearn library, we can decide what combinations of ngrams do we need to use: </h3>

(1,1) -> unigrams only

(2,2) -> bigrams only

(3,3) -> trigrams only

(1,2) -> unigrams and bigrams

(1,3) -> unigrams, bigrams, trigrams

(2,3) -> bigrams, trigrams and so on.
